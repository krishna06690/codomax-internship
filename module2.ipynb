```python

# MODULE 02: DATA ANALYSIS WITH PYTHON

# 1. Import libraries
import pandas as pd
import numpy as np

print("MODULE 02 - DATA ANALYSIS WITH PYTHON")
print("=" * 50)

# 2. Import CSV dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
data = pd.read_csv(url)

print("\nDataset imported successfully!")

# 3. Explore the dataset
print("\nFirst 5 rows:")
print(data.head())

print("\nDataset shape:")
print(data.shape)

print("\nDataset information:")
data.info()

# 4. Check missing values
print("\nMissing values before cleaning:")
print(data.isnull().sum())

# 5. Basic statistical analysis
print("\nSummary statistics:")
print(data.describe())

# 6. Clean the dataset
# Fill missing Age values with median
data["Age"] = data["Age"].fillna(data["Age"].median())

# Fill missing Embarked values with mode
data["Embarked"] = data["Embarked"].fillna(
    data["Embarked"].mode()[0]
)

# Remove Cabin because it contains many missing values
data.drop(columns=["Cabin"], inplace=True)

# Remove duplicate records
data.drop_duplicates(inplace=True)

print("\nMissing values after cleaning:")
print(data.isnull().sum())

# 7. Use NumPy for numerical analysis
average_age = np.mean(data["Age"])
print("\nAverage passenger age:", round(average_age, 2))

# Create survival status using NumPy
data["Survival_Status"] = np.where(
    data["Survived"] == 1, "Survived", "Not Survived"
)

print("\nSurvival status:")
print(data["Survival_Status"].value_counts())

# 8. Basic data analysis using Pandas
print("\nSurvival rate by gender (%):")
print(data.groupby("Sex")["Survived"].mean() * 100)

print("\nSurvival rate by passenger class (%):")
print(data.groupby("Pclass")["Survived"].mean() * 100)

print("\nAverage age by gender:")
print(data.groupby("Sex")["Age"].mean())

# 9. Final result
print("\nCleaned dataset:")
print(data.head())

print("\nMODULE 02 COMPLETED SUCCESSFULLY!")
```
