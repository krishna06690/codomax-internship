```python

# MODULE 03 - DATA VISUALIZATION WITH PYTHON

# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("MODULE 03 - DATA VISUALIZATION")
print("=" * 50)

# 2. Load the Iris dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
data = pd.read_csv(url)

print("\nDataset loaded successfully!")
print("\nFirst 5 rows:")
print(data.head())

print("\nDataset information:")
print(data.info())

# Set chart style
sns.set_theme(style="whitegrid")

# 3. BAR CHART
avg_sepal = data.groupby("species")["sepal_length"].mean()

plt.figure(figsize=(8, 5))
avg_sepal.plot(kind="bar", color=["skyblue", "orange", "green"])
plt.title("Average Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Sepal Length")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 4. LINE CHART
plt.figure(figsize=(8, 5))
plt.plot(data["sepal_length"], data["petal_length"],
         marker="o", color="purple", linewidth=1)
plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.tight_layout()
plt.show()

# 5. HISTOGRAM
plt.figure(figsize=(8, 5))
plt.hist(data["sepal_width"], bins=10,
         color="teal", edgecolor="black")
plt.title("Distribution of Sepal Width")
plt.xlabel("Sepal Width")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 6. SCATTER PLOT
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=data,
    x="sepal_width",
    y="petal_width",
    hue="species",
    s=70
)
plt.title("Sepal Width vs Petal Width by Species")
plt.xlabel("Sepal Width")
plt.ylabel("Petal Width")
plt.tight_layout()
plt.show()

# 7. PIE CHART
species_count = data["species"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    species_count,
    labels=species_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribution of Iris Species")
plt.tight_layout()
plt.show()

# 8. FINAL INSIGHTS
print("\nKEY DATA INSIGHTS")
print("-" * 40)

print("\nAverage sepal length:")
print(avg_sepal)

print("\nSpecies counts:")
print(species_count)

print("\nTotal records:", len(data))
print("\nAll five visualizations created successfully!")
print("MODULE 03 COMPLETED!")
```
