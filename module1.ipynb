```python

# =====================================================
# MODULE 1: PYTHON & DATA SCIENCE FUNDAMENTALS
# =====================================================

print("=" * 55)
print("MODULE 1: PYTHON & DATA SCIENCE FUNDAMENTALS")
print("=" * 55)

# 1. INTRODUCTION TO PYTHON
print("\n1. INTRODUCTION TO PYTHON")
print("Welcome to Python Programming!")
print("Learning Python for Data Science.")

# 2. VARIABLES AND DATA TYPES
print("\n2. VARIABLES AND DATA TYPES")

name = "Krishna"
age = 20
height = 5.8
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Student:", is_student)

print("Name Type:", type(name))
print("Age Type:", type(age))
print("Height Type:", type(height))
print("Student Type:", type(is_student))

# 3. PYTHON OPERATORS
print("\n3. PYTHON OPERATORS")

a = 20
b = 10

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
print("Exponent:", a ** 2)
print("Comparison:", a > b)

# 4. CONDITIONAL STATEMENTS
print("\n4. CONDITIONAL STATEMENTS")

marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")

# 5. LOOPS
print("\n5. LOOPS")

print("For Loop:")
for i in range(1, 6):
    print(i)

print("While Loop:")
count = 1
while count <= 5:
    print(count)
    count += 1

# 6. FUNCTIONS
print("\n6. FUNCTIONS")

def calculate_average(a, b, c):
    return (a + b + c) / 3

average = calculate_average(80, 90, 85)
print("Average Marks:", average)

# 7. LISTS
print("\n7. LISTS")

subjects = ["Python", "Mathematics", "Statistics", "Data Science"]

print("Subjects:", subjects)
print("First Subject:", subjects[0])

subjects.append("Machine Learning")
print("Updated Subjects:", subjects)

# 8. DICTIONARIES
print("\n8. DICTIONARIES")

student = {
    "name": "Krishna",
    "age": 20,
    "course": "BCA",
    "semester": 5
}

for key, value in student.items():
    print(key, ":", value)

# 9. FILE HANDLING
print("\n9. FILE HANDLING")

with open("student.txt", "w") as file:
    file.write("Python and Data Science Fundamentals")

with open("student.txt", "r") as file:
    content = file.read()

print("File Content:", content)

# 10. DATA SCIENCE USING PANDAS
print("\n10. DATA SCIENCE USING PANDAS")

import pandas as pd

data = {
    "Student": ["Aman", "Rahul", "Krishna", "Priya", "Rohit"],
    "Marks": [75, 85, 90, 80, 70],
    "Age": [20, 21, 20, 19, 22]
}

df = pd.DataFrame(data)

print("\nStudent Dataset:")
print(df)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nAverage Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

# 11. REAL-WORLD APPLICATIONS OF DATA SCIENCE
print("\n11. REAL-WORLD APPLICATIONS")

applications = [
    "Healthcare - Disease Prediction",
    "Banking - Fraud Detection",
    "E-commerce - Product Recommendations",
    "Education - Student Performance Analysis",
    "Transportation - Traffic Prediction"
]

for application in applications:
    print("-", application)

# 12. CONCLUSION
print("\n" + "=" * 55)
print("MODULE 1 COMPLETED SUCCESSFULLY!")
print("=" * 55)
print("Topics covered: Python Basics, Data Structures,")
print("Functions, File Handling and Data Science.")
```
